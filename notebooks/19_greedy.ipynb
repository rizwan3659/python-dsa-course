## 1. The greedy idea

A greedy algorithm builds the answer one step at a time, always taking the choice that looks best right now, and never goes back. It is fast and simple, and sometimes wrong.

- **Works: making change with 1, 5, 10, 25:** Take the largest coin that fits, repeat. 63 = 25 + 25 + 10 + 1 + 1 + 1. Optimal.
- **Fails: coins 1, 3, 4 and amount 6:** Greedy: 4 + 1 + 1 (3 coins). Best: 3 + 3 (2 coins). One counterexample is enough to reject greedy.

## 4. Sort, then decide

Most greedy algorithms are "sort by the right key, then one pass". The hard part is the key, and the exchange argument that proves it: swapping two neighbours out of greedy order never helps.

|---|---|---|

| Fractional knapsack | value / weight, descending | every unit of capacity gets the best value |
| Two city scheduling | cost A − cost B | the biggest savings go to A |
| Largest number | a + b vs b + a | the better of two neighbours goes first |
| Boats | weight (two pointers) | the heaviest person must go; pair with the lightest |

## Challenge: Maximum activities: try every subset → earliest end first

Choose the largest set of non-overlapping activities.

**Brute force** (O(2ⁿ)): In the worst case it checks every subset of the n activities.

```python
from itertools import combinations

def max_activities(intervals):
    for k in range(len(intervals), 0, -1):
        for group in combinations(sorted(intervals), k):
            if all(group[i][1] <= group[i + 1][0] for i in range(k - 1)):
                return k
    return 0
```

**Optimized** (O(n log n)): One sort and one pass.

Greedy replaces "try everything" with one provably safe choice per step. Always back the choice with an argument or a stress test.

```python
def max_activities(intervals):
    count, last_end = 0, float("-inf")
    for start, end in sorted(intervals, key=lambda x: x[1]):
        if start >= last_end:
            count += 1
            last_end = end
    return count
```
