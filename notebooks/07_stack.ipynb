## 1. LIFO: last in, first out

A stack only lets you touch the top. In Python, a list is a perfect stack: the end of the list is the top.

```text
         push(40)  ↓      ↑  pop() → 40
               ┌──────┐
       TOP  →  │  40  │
               ├──────┤
               │  30  │
               ├──────┤
               │  20  │
               ├──────┤
               │  10  │
               └──────┘
```

| Operation | Python | Cost |
|---|---|---|
| push | `stack.append(x)` | O(1) |
| pop | `stack.pop()` | O(1) |
| peek | `stack[-1]` | O(1) |
| is empty? | `not stack` | O(1) |

## 4. The call stack

## Challenge: Daily Temperatures: scan ahead → monotonic stack

For each day, how many days until a warmer temperature?

**Brute force** (O(n²)): A long cold spell makes every inner scan run far.

````python
def daily_temperatures(temps):
    answer = [0] * len(temps)
    for i in range(len(temps)):
        for j in range(i + 1, len(temps)):
            if temps[j] > temps[i]:
                answer[i] = j - i
                break
    return answer
````

**Optimized** (O(n)): Each index is pushed once and popped at most once.

Nested loops are not always O(n²): count how many times each element can be pushed and popped.

````python
def daily_temperatures(temps):
    answer = [0] * len(temps)
    stack = []
    for i, t in enumerate(temps):
        while stack and temps[stack[-1]] < t:
            j = stack.pop()
            answer[j] = i - j
        stack.append(i)
    return answer
````
