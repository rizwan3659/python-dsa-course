## 2. Building one

Each node holds a dict of children and an end flag. Insert creates missing children; search walks and checks the end flag; starts_with only walks.

**A compact trie**

````python
class TrieNode:
    def __init__(self):
        self.children = {}
        self.end = False

root = TrieNode()

def insert(word):
    node = root
    for ch in word:
        node = node.children.setdefault(ch, TrieNode())
    node.end = True

def walk(prefix):
    node = root
    for ch in prefix:
        node = node.children.get(ch)
        if node is None:
            return None
    return node

for w in ["cat", "car", "care"]:
    insert(w)
print(walk("car").end, walk("ca").end, walk("cab"))
````

## Challenge: Autocomplete: scan every word → walk a trie

After each typed character, suggest stored words that start with the typed prefix.

**Brute force** (O(n log n)): Each keystroke scans (and sorts) all n products.

````python
def suggestions(products, prefix):
    return sorted(p for p in products if p.startswith(prefix))[:3]
````

**Optimized** (O(L)): After building once, each keystroke costs O(1) and a query of length L costs O(L).

Pay once to organise the data by prefix, and every prefix query becomes a short walk.

````python
def build(products):
    root = {}
    for p in sorted(products):
        node = root
        for ch in p:
            node = node.setdefault(ch, {"#": []})
            if len(node["#"]) < 3:
                node["#"].append(p)
    return root
# each keystroke: node = node.get(ch); answer = node["#"]
````
