## Challenge: Koko Eating Bananas: try every speed → binary search on the answer

Find the minimum eating speed k that finishes all piles within h hours.

**Brute force** (O(n · M)): Up to M = max(piles) speeds, each checked in O(n). With piles of a billion bananas, hopeless.

````python
def min_eating_speed(piles, h):
    k = 1
    while sum((p + k - 1) // k for p in piles) > h:
        k += 1
    return k
````

**Optimized** (O(n log M)): About 30 checks for M up to a billion.

Binary search is not just for arrays: search any range of candidate answers with a monotonic check.

````python
def min_eating_speed(piles, h):
    lo, hi = 1, max(piles)
    while lo < hi:
        k = (lo + hi) // 2
        if sum((p + k - 1) // k for p in piles) <= h:
            hi = k
        else:
            lo = k + 1
    return lo
````
