## 1. Understand before you code

Most wrong answers in interviews come from solving a slightly different problem. Spend the first minutes making the problem precise.

- **Restate it:** Say the problem in your own words. What goes in, what comes out, of which type?
- **Examples:** Work one normal example by hand. Then invent your own, including weird ones.
- **Edge cases:** Empty input, one element, all equal, negatives, duplicates, huge values. Ask what should happen.

| If n can be up to… | Aim for | Typical idea |
|---|---|---|
| 10–20 | O(2ⁿ) or O(n!) | try every subset / order |
| 100–500 | O(n³) | three nested loops |
| 1,000–5,000 | O(n²) | all pairs |
| 10⁵–10⁶ | O(n log n) or O(n) | sort, hash map, two pointers |
| 10⁹ or more | O(log n) or O(1) | binary search, a formula |

Constraints are a hint: they tell you how fast the solution has to be. Python manages roughly 10⁷ simple steps per second.

## 3. Brute force first

A brute force tries every candidate answer. It is usually slow, but it is easy to get right, it proves you understand the problem, and it shows you where the wasted work is.

**From brute force to better**

- **List the candidates:** What could the answer be? Every pair? Every subarray? Every subset?
- **Check each one:** Write the simplest check. Correctness first.
- **Count the work:** How many candidates × cost per check? That is your complexity.
- **Find repeated work:** What do neighbouring candidates recompute? That is where a better data structure or pattern helps.

## 4. Best, average and worst cases

The same code can be fast on one input and slow on another. Big-O (upper bound), Big-Omega (lower bound) and Big-Theta (both) let you say precisely how fast, and for which case.

| Symbol | Means | Linear search example |
|---|---|---|
| O(f) | grows **at most** like f | worst case is O(n) |
| Ω(f) | grows **at least** like f | best case is Ω(1); every case is Ω(1) |
| Θ(f) | grows **exactly** like f (both) | worst case is Θ(n), best case is Θ(1) |

When people say "linear search is O(n)" they mean the worst case. Interviews almost always ask about the worst case.

## 5. Math every programmer uses

A little math turns loops into formulas and O(n) into O(√n) or O(log n). You only need a handful of facts.

- **Sums:** 1 + 2 + … + n = n(n + 1) / 2. Number of pairs = n(n − 1) / 2.
- **Logarithms:** Halving n until it reaches 1 takes log₂ n steps. Counting digits takes log₁₀ n steps.
- **Divisors:** Divisors come in pairs (d, n / d), so checking up to √n is enough.
- **Modulo:** `x % m` wraps around into 0..m − 1: clocks, circular arrays, even/odd, last digit.
- **GCD:** Euclid: gcd(a, b) = gcd(b, a % b). Fast: O(log min(a, b)).
- **Integer division:** Use `//` for whole numbers. `/` always gives a float and can lose precision.

## Challenge: Count pairs with a given sum: all pairs → dictionary

Count pairs i < j with nums[i] + nums[j] == target.

**Brute force** (O(n²)): Every pair is checked: n(n − 1) / 2 checks.

````python
def count_pairs(nums, target):
    count = 0
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                count += 1
    return count
````

**Optimized** (O(n)): One pass, one dictionary lookup and one update per element.

Write the brute force, find the question the inner loop keeps re-asking, then answer it with a data structure.

````python
def count_pairs(nums, target):
    seen = {}
    count = 0
    for x in nums:
        count += seen.get(target - x, 0)
        seen[x] = seen.get(x, 0) + 1
    return count
````
