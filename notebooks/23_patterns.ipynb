## 1. How to recognise a pattern

Most interview problems are one of about a dozen patterns in disguise. Learn the signals, and the first minute of every problem becomes a lookup.

| If the problem says… | Think… |
|---|---|
| sorted array, pair / triplet, in place | Two pointers |
| contiguous subarray / substring, longest / shortest / k in a row | Sliding window |
| sum of a range, many queries, subarray sum = k | Prefix sum (+ hash map) |
| seen before, count, group, complement | Hashing |
| linked list middle / cycle / k-th from end | Fast and slow pointers |
| sorted, O(log n), minimum value that works | Binary search (on index or answer) |
| next greater / smaller, days until, span | Monotonic stack |
| fewest steps, levels, nearest, spreading | BFS |
| explore everything, connected regions, paths in trees | DFS |
| all combinations / permutations / subsets | Backtracking |
| k largest / most frequent, merge k sorted, running median | Heap |
| meetings, ranges, overlapping | Intervals (sort first) |
| locally best choice is provably globally best | Greedy |
| number of ways, min/max cost, overlapping choices | Dynamic programming |
| appears once / twice, subsets of n ≤ 20, XOR | Bit manipulation |
| range sums or minimums with updates | Fenwick / segment tree |

## 2. Two pointers

Two indexes walk the data together: from both ends toward the middle, or a slow writer behind a fast reader.

- **When to use it:** Sorted arrays, pairs or triplets with a target, palindromes, and in-place rearranging with O(1) extra space.
- **How to recognise it:** - “sorted”
- “pair / triplet that sums to”
- “in place”, “O(1) extra space”
- “palindrome”
- **Typical complexity:** O(n) time (O(n²) for 3Sum), O(1) space.

**Recognise it** (answer these in the browser course or locally):

- Pair in a sorted array (choose, `pt-r1`)
- In-place filter (choose, `pt-r28`)

**Visualizer:** Two pointers: open https://rizwan3659.github.io/python-dsa-course/#v/two-pointers

**Code template**

```python
left, right = 0, len(a) - 1
while left < right:
    if condition(a[left], a[right]):
        return answer
    if need_bigger:
        left += 1
    else:
        right -= 1
```

## 3. Sliding window

A window [left, right] slides over the data. Update it incrementally instead of recomputing it.

- **When to use it:** Contiguous subarrays or substrings: maximum/minimum/longest/shortest with some condition.
- **How to recognise it:** - “contiguous”, “subarray”, “substring”
- “k consecutive”
- “longest / shortest such that …”
- **Typical complexity:** O(n): left and right each move at most n times.

**Recognise it** (answer these in the browser course or locally):

- Best run of k days (choose, `pt-r2`)

**Visualizer:** Sliding window: open https://rizwan3659.github.io/python-dsa-course/#v/sliding-window

```python
left = 0
```

## 4. Prefix sum

Store running totals once; any range sum becomes a subtraction. Add a hash map of prefix sums to count subarrays with a given sum.

- **When to use it:** Many range-sum queries, or “subarray sums to k” when numbers can be negative.
- **How to recognise it:** - “sum of elements from i to j”
- “number of subarrays with sum k”
- “pivot / balance point”
- **Typical complexity:** O(n) to build, O(1) per query.

**Recognise it** (answer these in the browser course or locally):

- Many range sums (choose, `pt-r4`)
- Subarrays summing to k (with negatives) (choose, `pt-r15`)

**Visualizer:** Prefix sums: open https://rizwan3659.github.io/python-dsa-course/#v/prefix-sum

**Code template**

```python
prefix = [0]
for x in a:
    prefix.append(prefix[-1] + x)
# sum(a[l..r]) = prefix[r + 1] - prefix[l]

seen, total, count = {0: 1}, 0, 0      # subarrays summing to k
for x in a:
    total += x
    count += seen.get(total - k, 0)
    seen[total] = seen.get(total, 0) + 1
```

## 5. Hashing

Trade memory for time: a set or dict answers “seen it?”, “how many?” and “where?” in O(1).

- **When to use it:** Whenever an inner loop searches for a specific value, or you need counts or groups.
- **How to recognise it:** - “duplicate”, “seen”, “unique”
- “two numbers that add up to”
- “anagram”, “frequency”, “group”
- **Typical complexity:** O(n) time, O(n) space.

**Recognise it** (answer these in the browser course or locally):

- Seen before? (choose, `pt-r3`)
- Grouping words (choose, `pt-r16`)

**Visualizer:** Two Sum with a hash map: open https://rizwan3659.github.io/python-dsa-course/#v/two-sum-hash

**Code template**

```python
seen = {}
for i, x in enumerate(a):
    need = target - x
    if need in seen:
        return [seen[need], i]
    seen[x] = i
```

## 6. Fast and slow pointers

Two pointers at different speeds: find the middle, detect cycles, locate the cycle start, all in O(1) space.

- **When to use it:** Linked lists (middle, cycle, k-th from end) and any “next value” function that might loop.
- **How to recognise it:** - “linked list” + “middle” or “cycle”
- “without extra space”
- “happy number”-style repeating sequences
- **Typical complexity:** O(n) time, O(1) space.

**Recognise it** (answer these in the browser course or locally):

- Middle of a linked list (choose, `pt-r5`)
- Is there a loop? (choose, `pt-r17`)

**Visualizer:** Fast and slow pointers: open https://rizwan3659.github.io/python-dsa-course/#v/fast-slow

**Code template**

```python
slow = fast = head
while fast and fast.next:
    slow = slow.next
    fast = fast.next.next
    if slow is fast:
        return True        # cycle
return False               # (for the middle: return slow)
```

## 7. Binary search

Halve the search space every step. Works on sorted arrays and on any monotonic yes/no question about the answer.

- **When to use it:** Sorted data, O(log n) requirements, and “minimum / maximum value such that …”.
- **How to recognise it:** - “sorted” or “rotated sorted”
- “O(log n)”
- “smallest speed / capacity / time that works”
- **Typical complexity:** O(log n) steps (× the cost of each check).

**Recognise it** (answer these in the browser course or locally):

- Minimum speed (choose, `pt-r6`)
- Rotated array (choose, `pt-r18`)
- Split into k parts (choose, `pt-r36`)

**Visualizer:** Binary search: open https://rizwan3659.github.io/python-dsa-course/#v/binary-search

**Code template**

```python
lo, hi = low_bound, high_bound
while lo < hi:
    mid = (lo + hi) // 2
    if works(mid):
        hi = mid           # mid might be the answer
    else:
        lo = mid + 1
return lo
```

## 8. Monotonic stack

Keep a stack that is always increasing (or decreasing). When a new element breaks the order, everything it pops has just found its answer.

- **When to use it:** Next/previous greater or smaller element, spans, histograms.
- **How to recognise it:** - “next greater”, “next warmer”
- “how many days until”
- “largest rectangle”
- **Typical complexity:** O(n): each element is pushed and popped at most once.

**Recognise it** (answer these in the browser course or locally):

- Next warmer day (choose, `pt-r7`)
- Largest rectangle (choose, `pt-r19`)

**Visualizer:** Monotonic stack: next greater element: open https://rizwan3659.github.io/python-dsa-course/#v/monotonic-stack

**Code template**

```python
result = [-1] * len(a)
stack = []                         # indexes
for i, x in enumerate(a):
    while stack and a[stack[-1]] < x:
        result[stack.pop()] = x
    stack.append(i)
```

## 9. Breadth-first search

Explore in rings of distance with a queue. First arrival = fewest steps.

- **When to use it:** Shortest paths in unweighted graphs and grids, level-by-level tree problems, spreading processes.
- **How to recognise it:** - “minimum number of steps / moves”
- “level”, “nearest”
- “every minute it spreads”
- **Typical complexity:** O(V + E) (grids: O(rows × cols)).

**Recognise it** (answer these in the browser course or locally):

- Fewest moves (choose, `pt-r8`)
- Spreading process (choose, `pt-r20`)
- Tree level averages (choose, `pt-r30`)

**Visualizer:** Grids as graphs: count islands: open https://rizwan3659.github.io/python-dsa-course/#v/grid-islands

**Code template**

```python
from collections import deque
queue, seen = deque([start]), {start}
while queue:
    u = queue.popleft()
    for v in neighbours(u):
        if v not in seen:
            seen.add(v)
            queue.append(v)
```

## 10. Depth-first search

Go deep, then backtrack, with recursion or an explicit stack. The natural fit for trees, connected regions and exhaustive exploration.

- **When to use it:** Trees (heights, paths, validations), connected components, flood fill, reachability.
- **How to recognise it:** - “tree” (almost always)
- “connected”, “region”, “island”
- “all paths”
- **Typical complexity:** O(V + E), recursion depth = height.

**Recognise it** (answer these in the browser course or locally):

## 11. Backtracking

Build candidates step by step; when one cannot lead to a solution, undo the last choice and try another.

- **When to use it:** Generate all combinations, permutations, subsets, or placements under constraints.
- **How to recognise it:** - “all”, “every”, “generate”, “list the ways”
- small input sizes (n ≤ 20)
- constraints like “no two attack”
- **Typical complexity:** Exponential (2ⁿ, n!) but pruned.

**Recognise it** (answer these in the browser course or locally):

- All combinations (choose, `pt-r10`)
- Place queens (choose, `pt-r25`)

**Visualizer:** Backtracking: generate all subsets: open https://rizwan3659.github.io/python-dsa-course/#v/backtracking

**Code template**

```python
def backtrack(path, start):
    if done(path):
        result.append(path[:])
        return
    for choice in options(start):
        path.append(choice)        # choose
        backtrack(path, next_start)
        path.pop()                 # un-choose
```

## 12. Heap / top-K

A heap hands you the smallest (or largest) item in O(log n). Keep a heap of size k for top-K questions.

- **When to use it:** k largest/smallest/most frequent, merging sorted streams, scheduling, running medians.
- **How to recognise it:** - “k largest / closest / most frequent”
- “merge k sorted”
- “median of a stream”
- **Typical complexity:** O(n log k).

**Recognise it** (answer these in the browser course or locally):

- K most frequent (choose, `pt-r11`)
- Running median (choose, `pt-r24`)

**Visualizer:** Heap: push, pop, bubble up and down: open https://rizwan3659.github.io/python-dsa-course/#v/heap

**Code template**

```python
import heapq
h = []
for x in items:
    heapq.heappush(h, x)
    if len(h) > k:
        heapq.heappop(h)      # drop the smallest
return h[0]                    # k-th largest
```

## 13. Intervals

Sort by start (or end), then sweep once. Overlap means: next start ≤ current end.

- **When to use it:** Meetings, bookings, ranges, merging or counting overlaps.
- **How to recognise it:** - pairs [start, end]
- “overlap”, “merge”, “rooms”, “free time”
- **Typical complexity:** O(n log n) for the sort.

**Recognise it** (answer these in the browser course or locally):

**Visualizer:** Intervals: sort, then merge: open https://rizwan3659.github.io/python-dsa-course/#v/intervals

**Code template**

```python
merged = []
for start, end in sorted(intervals):
    if merged and start <= merged[-1][1]:
        merged[-1][1] = max(merged[-1][1], end)
```

## 14. Greedy

Make the locally best choice at each step and never undo it. It works only when you can argue that a best choice never hurts later ones.

- **When to use it:** Scheduling by earliest end, reachability, coin-like exchanges with nice denominations, matching sorted lists.
- **How to recognise it:** - “minimum number of …” with a simple local rule
- “can you reach”
- sorting makes the choice obvious
- **Typical complexity:** Often O(n) or O(n log n).

**Recognise it** (answer these in the browser course or locally):

## 15. Combining patterns

Hard interview problems usually glue two or three patterns together: prefix sums plus a monotonic deque, sorting plus a heap, binary search plus a greedy check. Name each piece before you write code.

| Problem | Pieces |
|---|---|
| Shortest subarray with sum ≥ k (negatives allowed) | prefix sums + monotonic deque |
| IPO / refueling stops | sort + max-heap + greedy |
| Split array largest sum | binary search on the answer + greedy check |
| Minimum window substring | sliding window + counting map |
| Count smaller after self | coordinate compression + Fenwick tree |
| Word ladder | BFS over an implicit graph + hashing |

> **Routine for a hard problem:** brute force first (say its cost) → find the repeated work → name the pattern that removes it → check whether a second pattern is needed for what remains.
