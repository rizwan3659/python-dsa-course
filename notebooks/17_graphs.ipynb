Nodes (vertices) connected by edges. Edges can be one-way (directed) or two-way (undirected), and can carry weights. Trees, grids, maps and social networks are all graphs.

```text
  undirected            directed              weighted
  A — B                 A → B                 A —4— B
  |   |                 ↑   ↓                 |2    |1
  C — D                 C ← D                 C —3— D
```

## 2. Storing a graph

An adjacency list stores each node’s neighbours: compact for sparse graphs. An adjacency matrix stores a V × V grid of 0/1 (or weights): instant edge checks, but V² memory.

**Both representations**

````python
edges = [("A", "B"), ("A", "C"), ("B", "D"), ("C", "D")]

adj = {}                                   # adjacency list
for u, v in edges:
    adj.setdefault(u, []).append(v)
    adj.setdefault(v, []).append(u)
print(adj)

names = ["A", "B", "C", "D"]                # adjacency matrix
idx = {n: i for i, n in enumerate(names)}
m = [[0] * len(names) for _ in names]
for u, v in edges:
    m[idx[u]][idx[v]] = m[idx[v]][idx[u]] = 1
for row in m:
    print(row)
````

## Challenge: Distance to the Nearest 0: BFS per cell → one multi-source BFS

For every cell of a 0/1 grid, find the distance to the nearest 0.

**Brute force** (O(n²)): With n = R × C cells, a full BFS for each cell costs O(n) each: O(n²) total.

````python
def update_matrix(mat):
    rows, cols = len(mat), len(mat[0])
    out = [[0] * cols for _ in range(rows)]
    for r in range(rows):
        for c in range(cols):
            if mat[r][c] == 1:
                out[r][c] = bfs_to_nearest_zero(mat, r, c)   # O(R·C) each
    return out
````

**Optimized** (O(n)): Every cell enters the queue once.

When many searches share the same goal, reverse the direction and search once from all the goals.

````python
from collections import deque

def update_matrix(mat):
    rows, cols = len(mat), len(mat[0])
    dist = [[-1] * cols for _ in range(rows)]
    q = deque()
    for r in range(rows):
        for c in range(cols):
            if mat[r][c] == 0:
                dist[r][c] = 0
                q.append((r, c))
    while q:
        r, c = q.popleft()
        for nr, nc in ((r + 1, c), (r - 1, c), (r, c + 1), (r, c - 1)):
            if 0 <= nr < rows and 0 <= nc < cols and dist[nr][nc] == -1:
                dist[nr][nc] = dist[r][c] + 1
                q.append((nr, nc))
    return dist
````
