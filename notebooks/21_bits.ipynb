## 1. Numbers in binary

```text

 13 =          1    1    0    1     8 + 4 + 1 = 13
```

**Binary in Python**

````python
print(bin(13), int("1101", 2), f"{13:08b}")
print((13).bit_length(), bin(13).count("1"))
````

## 2. The bitwise operators

&, | and ^ combine two numbers column by column. > slide all the bits. ~ flips them (and in Python, ~x is −x − 1).

**Visualizer:** Bitwise operators, bit by bit: open https://rizwan3659.github.io/python-dsa-course/#v/bitwise

Try AND, OR and XOR on 44 and 27, then the shifts.

| Operator | Rule per bit | Typical use |
|---|---|---|
| a & b | 1 if both are 1 | test or clear bits (masking) |
| a \| b | 1 if either is 1 | set bits |
| a ^ b | 1 if they differ | toggle bits, cancel pairs |
| ~a | flip every bit | build "all except" masks |
| a << k | shift up k places | multiply by 2ᵏ, build masks 1 << i |
| a >> k | shift down k places | floor-divide by 2ᵏ, read bit i |

Precedence in Python: + and − bind tighter than shifts, shifts tighter than &, then ^, then |, and comparisons come last. So 1 << i - 1 is 1 << (i − 1), and x & 1 == 1 is (x & 1) == 1. (C and Java differ!) When in doubt, add brackets.

## 3. Masks: get, set, clear, toggle

A mask is a number with 1s exactly where you want to act. 1 << i selects bit i; combine it with the right operator to read or change that bit.

**The four operations**

````python
n, i = 0b1010, 0
print((n >> i) & 1)        # get   -> 0
print(n | (1 << i))        # set   -> 11
print(n & ~(1 << 1))       # clear -> 8
print(n ^ (1 << 3))        # toggle -> 2
````

## 5. Subsets as bitmasks

With n items, the numbers 0 .. 2ⁿ − 1 are exactly the 2ⁿ subsets: bit i says whether item i is in. This replaces recursion for small n and is the basis of bitmask DP.

```text
 items = [a, b, c]
 mask  binary  subset
  0     000    []
  1     001    [a]
  2     010    [b]
  3     011    [a, b]
  4     100    [c]
  ...
  7     111    [a, b, c]
```

## Challenge: Single number: counting → XOR

Every value appears twice except one. Find it.

**Brute force** (O(n²)): count() scans the whole list for every element.

````python
def single_number(nums):
    for x in nums:
        if nums.count(x) == 1:
            return x
````

**Optimized** (O(n)): One pass, one integer of extra memory.

When values come in pairs, XOR can cancel them for free.

````python
def single_number(nums):
    result = 0
    for x in nums:
        result ^= x
    return result
````
