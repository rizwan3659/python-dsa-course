## 1. What a heap is

A heap is a complete binary tree where every parent is ≤ its children (min-heap) or ≥ them (max-heap). So the smallest (or largest) value is always at the root.

```text
          2                index:  0  1  2  3  4
         / \               list:  [2, 5, 4, 9, 7]
        5   4
       / \                 children of i: 2i + 1, 2i + 2
      9   7                parent of i:   (i - 1) // 2
```

## 3. heapq in Python

Python’s heapq module turns a plain list into a min-heap. For a max-heap, push negated values; for priorities, push tuples.

**The essentials**

````python
import heapq

h = []
for x in [5, 1, 8, 3]:
    heapq.heappush(h, x)        # O(log n)
print(h[0])                     # smallest, O(1)
print(heapq.heappop(h))         # 1

nums = [9, 4, 7]
heapq.heapify(nums)             # O(n), in place
print(heapq.nsmallest(2, [5, 2, 8, 1]), heapq.nlargest(2, [5, 2, 8, 1]))

maxh = [-x for x in [3, 9, 4]]  # max-heap trick
heapq.heapify(maxh)
print(-maxh[0])
````

## Challenge: Kth Largest: sort → heap of size k

Return the k-th largest element of a list of n numbers (k is small, n is huge).

**Brute force** (O(n log n)): Sorts everything, although we only care about k values.

````python
def find_kth_largest(nums, k):
    return sorted(nums, reverse=True)[k - 1]
````

**Optimized** (O(n log k)): Each step costs log k; memory is O(k).

For “top k”, maintain a heap of size k instead of sorting everything.

````python
import heapq

def find_kth_largest(nums, k):
    h = []
    for x in nums:
        heapq.heappush(h, x)
        if len(h) > k:
            heapq.heappop(h)
    return h[0]
````
