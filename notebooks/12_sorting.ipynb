## 4. Sorting in Python

In real code you call sorted() or list.sort(): Timsort, stable, O(n log n). The skill is choosing the key.

**Keys and stability**

````python
people = [("ann", 30), ("bob", 25), ("cat", 30)]
print(sorted(people, key=lambda p: p[1]))            # by age; ties keep their order
print(sorted(people, key=lambda p: (-p[1], p[0])))   # age desc, then name
words = ["banana", "Apple", "cherry"]
print(sorted(words, key=str.lower))                  # case-insensitive
print(sorted(words, key=len, reverse=True))
````

## Challenge: Count Inversions: every pair → merge sort

Count pairs i < j with nums[i] > nums[j].

**Brute force** (O(n²)): Every pair is checked.

````python
def count_inversions(nums):
    count = 0
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] > nums[j]:
                count += 1
    return count
````

**Optimized** (O(n log n)): Same cost as merge sort itself.

Sorting algorithms are also tools: piggy-back extra counting on their work.

````python
def count_inversions(nums):
    def sort(a):
        if len(a) <= 1:
            return a, 0
        mid = len(a) // 2
        left, x = sort(a[:mid])
        right, y = sort(a[mid:])
        out, i, j, inv = [], 0, 0, x + y
        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                out.append(left[i]); i += 1
            else:
                out.append(right[j]); j += 1
                inv += len(left) - i
        return out + left[i:] + right[j:], inv
    return sort(nums)[1]
````

| Question | Answer |
|---|---|
| Bubble sort? | Swap neighbours; O(n²); O(n) best with early exit. |
| Selection sort? | Pick the minimum each pass; always O(n²); not stable. |
| Insertion sort? | Grow a sorted prefix; great on nearly sorted data. |
| Merge sort? | Split, sort halves, merge; O(n log n), O(n) space, stable. |
| Quick sort? | Partition around a pivot; O(n log n) expected, O(n²) worst. |
| Heap sort? | Build a max-heap, extract max repeatedly; O(n log n), O(1) space. |
| Stable? | Equal keys keep their original order. |
| Python sort? | Timsort, stable; use key=. |
