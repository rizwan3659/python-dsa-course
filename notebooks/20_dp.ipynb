## 1. From recursion to memoization

Plain recursion often solves the same subproblem over and over. Cache each answer the first time, and an exponential algorithm becomes linear.

**Two ways to memoize**

```python
from functools import lru_cache

@lru_cache(maxsize=None)          # 1) let Python cache the results
def fib(n):
    return n if n < 2 else fib(n - 1) + fib(n - 2)

def fib2(n, memo={0: 0, 1: 1}):   # 2) a dict you manage yourself
    if n not in memo:
        memo[n] = fib2(n - 1) + fib2(n - 2)
    return memo[n]

print(fib(80), fib2(80))
```

A mutable default like `memo={}` is shared between calls. That is usually a bug (topic 1), but here it is a deliberate cache.

## 2. Tabulation: state, base case, transition

Bottom-up DP fills a table from the smallest subproblems upwards. Every DP solution is the same four decisions: what a cell means, the base cases, how a cell is computed from smaller ones, and the order to fill them.

**The DP recipe**

- **State:** Say exactly what `dp[i]` means, in words. "dp[i] = the most money from houses 0..i."
- **Base cases:** The smallest states you can answer directly: dp[0], an empty prefix, the first row.
- **Transition:** How dp[i] follows from smaller states. Think about the last decision: "rob house i or not".
- **Order and answer:** Fill so that every dependency is ready first. Which cell holds the final answer?

## Challenge: Coin change: recursion → table

Fewest coins that make the amount, or −1.

**Brute force** (O(2ⁿ)): Each call branches on every coin, and the same amounts are solved again and again: exponential in the amount.

```python
def coin_change(coins, amount):
    if amount == 0:
        return 0
    best = float("inf")
    for c in coins:
        if c <= amount:
            sub = coin_change(coins, amount - c)
            if sub != -1:
                best = min(best, sub + 1)
    return best if best != float("inf") else -1
```

**Optimized** (O(n²)): O(amount × k): quadratic-style "states × choices". (Pick O(n²) as the closest option.)

Count the distinct states. If there are few states but many calls, cache them: that is dynamic programming.

```python
def coin_change(coins, amount):
    dp = [0] + [float("inf")] * amount
    for a in range(1, amount + 1):
        for c in coins:
            if c <= a:
                dp[a] = min(dp[a], dp[a - c] + 1)
    return dp[amount] if dp[amount] != float("inf") else -1
```
