## 1. Strings are immutable sequences

```text
      -6-5-4-3-2-1         s[::-1] → "nohtyp"
```

## Challenge: Longest Substring Without Repeats: restart everywhere → sliding window

Return the length of the longest substring with no repeated characters.

**Brute force** (O(n²)): Every start index re-scans forward.

````python
def length_of_longest_substring(s):
    best = 0
    for i in range(len(s)):
        seen = set()
        for j in range(i, len(s)):
            if s[j] in seen:
                break
            seen.add(s[j])
            best = max(best, j - i + 1)
    return best
````

**Optimized** (O(n)): One pass; the dict lookups are O(1).

When a window becomes invalid, repair it by moving its left edge instead of starting over.

````python
def length_of_longest_substring(s):
    last, left, best = {}, 0, 0
    for right, ch in enumerate(s):
        if ch in last and last[ch] >= left:
            left = last[ch] + 1
        last[ch] = right
        best = max(best, right - left + 1)
    return best
````
