## 2. Why list.pop(0) is a trap

A list stores items side by side, so removing the first one shifts all the others. collections.deque is built for O(1) work at both ends.

| Operation | list | deque |
|---|---|---|
| add at rear | `append` O(1) | `append` O(1) |
| remove from front | `pop(0)` **O(n)** | `popleft` O(1) |
| add at front | `insert(0, x)` **O(n)** | `appendleft` O(1) |
| index the middle | O(1) | **O(n)** |

## Challenge: Sliding Window Maximum: max of every window → monotonic deque

Return the maximum of every window of size k.

**Brute force** (O(n · k)): Each of about n windows scans k numbers.

````python
def max_sliding_window(nums, k):
    return [max(nums[i:i + k]) for i in range(len(nums) - k + 1)]
````

**Optimized** (O(n)): Each index enters and leaves the deque once.

Discard candidates that can never win. A deque lets you discard from both ends in O(1).

````python
from collections import deque

def max_sliding_window(nums, k):
    dq, out = deque(), []
    for i, x in enumerate(nums):
        while dq and nums[dq[-1]] <= x:
            dq.pop()
        dq.append(i)
        if dq[0] <= i - k:
            dq.popleft()
        if i >= k - 1:
            out.append(nums[dq[0]])
    return out
````
