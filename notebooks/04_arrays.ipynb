## 1. Arrays and indexing

A Python list is a dynamic array: its elements sit side by side in memory, so any position can be reached in one jump.

```text
 index:      0      1      2      3      4
          ┌──────┬──────┬──────┬──────┬──────┐
  nums =  │  10  │  20  │  30  │  40  │  50  │
          └──────┴──────┴──────┴──────┴──────┘
 index:     -5     -4     -3     -2     -1
```

Address of nums[i] = start + i × (size of one slot). That is why reading or writing nums[i] is O(1).

- **Positive index:** `nums[0]` is the first item, `nums[len(nums) - 1]` the last.
- **Negative index:** `nums[-1]` is the last item, `nums[-2]` the one before it.
- **Out of range:** `nums[5]` on a 5-item list raises `IndexError`. Slices like `nums[2:99]` never raise.

**Try it**

````python
nums = [10, 20, 30, 40, 50]
print(nums[0], nums[-1])       # 10 50
print(nums[1:4])               # [20, 30, 40]
nums[2] = 99                   # O(1) update
print(nums, len(nums))
````

## 2. Traversal: visiting every element

Most array algorithms are a loop plus a little state: a total, a best-so-far, a count. Learn to trace that state by hand.

**Four ways to loop**

````python
nums = [4, 7, 2]
for x in nums:                  # values only
    print(x)
for i in range(len(nums)):      # indexes (needed to modify nums[i])
    nums[i] *= 10
for i, x in enumerate(nums):    # index and value
    print(i, x)
for x in reversed(nums):        # backwards, no copy
    print(x)
````

## 3. Insert, delete, search, update

Updating or reading by index is instant. Inserting or deleting in the middle is not: every element after the gap has to shift.

**Visualizer:** Array operations: open https://rizwan3659.github.io/python-dsa-course/#v/array-ops

Insert 99 at index 0, then at the end. Count the shifts. Then delete index 1 and search for a value that is not there.

| Operation | Python | Cost | Why |
|---|---|---|---|
| Read / update by index | `a[i]`, `a[i] = x` | O(1) | direct address |
| Add / remove at the end | `a.append(x)`, `a.pop()` | O(1)* | no shifting (*amortized) |
| Insert / delete at i | `a.insert(i, x)`, `del a[i]`, `a.pop(i)` | O(n) | n − i elements shift |
| Search by value | `x in a`, `a.index(x)`, `a.remove(x)` | O(n) | scans from the start |
| Copy a slice | `a[i:j]` | O(j − i) | builds a new list |

## 4. Dynamic arrays and amortized cost

A list keeps spare capacity. When it fills up, Python allocates a bigger block and copies everything over. That copy is O(n), but it is rare.

```text
 capacity 4, full:    [10][20][30][40]
 append(50):          allocate 8 slots, copy 4 elements, then write 50
                      [10][20][30][40][50][  ][  ][  ]
 next 3 appends:      no copying at all
```

> Because capacity grows by a constant factor, n appends cost about 2n element copies in total. Spread over n appends that is O(1) each: **amortized O(1)**.

## 5. Two pointers

Two indexes that walk the array together. Opposite ends moving inward (reverse, pair sum in sorted data) or a slow writer chasing a fast reader (remove duplicates, move zeroes).

- **left → … ← right:** `left, right = 0, len(a) - 1
while left < right:
    ...
    left += 1  # and/or
    right -= 1`Reverse, palindrome check, pair sum in a sorted array, container with most water.
- **write chases read:** `write = 0
for read in range(len(a)):
    if keep(a[read]):
        a[write] = a[read]
        write += 1`Remove duplicates, move zeroes, filter in place with O(1) extra space.

## 7. Sliding window

For questions about contiguous ranges (“subarray”, “substring”, “k in a row”), keep a window and update it as it moves instead of recomputing it.

- **Add one, drop one:** `window = sum(a[:k])
for right in range(k, len(a)):
    window += a[right] - a[right - k]`
- **Grow right, shrink left:** `left = 0
for right in range(len(a)):
    add(a[right])
    while window_is_invalid():
        remove(a[left])
        left += 1
    record(right - left + 1)`

## 8. Kadane's algorithm

Maximum subarray sum in one pass. At each number ask one question: extend the run that ends just before it, or start fresh here?

**The whole algorithm**

````python
def max_subarray(nums):
    cur = best = nums[0]
    for x in nums[1:]:
        cur = max(x, cur + x)     # extend or restart
        best = max(best, cur)     # remember the best run
    return best

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))   # 6
````

## Challenge: Two Sum: nested loops → hash map

Return the indices of the two numbers in `nums` that add up to `target`.

`two_sum([2, 7, 11, 15], 9)` → `[0, 1]`

**Brute force** (O(n²)): About n²/2 pairs. For 100,000 numbers that is 5 billion checks.

````python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]
````

**Optimized** (O(n)): One loop with O(1) work per element. Space is O(n) for the dict: we traded memory for time.

When an inner loop is **searching for a specific value**, replace it with a set or dict lookup. This one idea turns dozens of O(n²) problems into O(n).

````python
def two_sum(nums, target):
    seen = {}                      # value -> index
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
````

## Challenge: Maximum Subarray: all ranges → Kadane

Return the largest sum of any non-empty contiguous subarray.

**Brute force** (O(n²)): n starts × up to n ends.

````python
def max_subarray(nums):
    best = nums[0]
    for i in range(len(nums)):
        total = 0
        for j in range(i, len(nums)):
            total += nums[j]
            best = max(best, total)
    return best
````

**Optimized** (O(n)): One pass, O(1) work per element, O(1) space.

Ask “what is the best answer **ending here**?” and build it from the previous one. That is the seed of dynamic programming.

````python
def max_subarray(nums):
    cur = best = nums[0]
    for x in nums[1:]:
        cur = max(x, cur + x)
        best = max(best, cur)
    return best
````

## Challenge: Container With Most Water: all pairs → two pointers

Pick two walls; the water they hold is `min(height) × distance`. Return the maximum.

**Brute force** (O(n²)): Every pair of walls.

````python
def max_area(height):
    best = 0
    for i in range(len(height)):
        for j in range(i + 1, len(height)):
            best = max(best, min(height[i], height[j]) * (j - i))
    return best
````

**Optimized** (O(n)): Each iteration moves one pointer; they meet after n − 1 moves.

Two pointers work when each comparison lets you **throw away** one candidate for good.

````python
def max_area(height):
    left, right = 0, len(height) - 1
    best = 0
    while left < right:
        best = max(best, min(height[left], height[right]) * (right - left))
        if height[left] < height[right]:
            left += 1
        else:
            right -= 1
    return best
````
