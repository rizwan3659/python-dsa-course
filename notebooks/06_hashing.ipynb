## 1. Why hashing is fast

A list must be scanned to find a value. A hash table computes where the value lives: the key goes through a hash function, which picks a bucket.

```text
      key            "cat"
       │
       ▼
    hash()     →   312            (a number computed from the key)
       │
       ▼
  312 % 8      →   bucket 0       (8 buckets in the table)
       │
       ▼
    bucket 0   →   ("cat", 3)     (the stored key-value pair)
```

## 2. Collisions and resizing

Different keys can land in the same bucket. Each bucket keeps a short chain, and the table grows when chains would get long.

- **Collision:** Two different keys, same bucket. Unavoidable: there are far more possible keys than buckets.
- **Chaining:** A bucket holds a small list of (key, value) pairs. Lookup compares keys within that one list.
- **Resizing:** When keys per bucket (the **load factor**) passes a limit, allocate ~2× buckets and re-insert everything. Rare, so still O(1) on average.

## 4. Counter and defaultdict

Two dict subclasses from collections that remove boilerplate: Counter counts, defaultdict creates missing values for you.

**Counter and defaultdict**

````python
from collections import Counter, defaultdict

c = Counter("banana")
print(c)                       # Counter({'a': 3, 'n': 2, 'b': 1})
print(c["z"])                  # 0, no KeyError
print(c.most_common(1))        # [('a', 3)]

groups = defaultdict(list)
for word in ["apple", "avocado", "banana"]:
    groups[word[0]].append(word)
print(dict(groups))            # {'a': ['apple', 'avocado'], 'b': ['banana']}
````

## 5. Hashing patterns

Five ideas cover most hashing problems: seen-set, complement lookup, counting, grouping by a key, and prefix sum + map.

| Pattern | Signal in the problem | Core line |
|---|---|---|
| Seen set | “duplicate”, “first repeat”, “cycle” | `if x in seen: …; seen.add(x)` |
| Complement lookup | “two numbers that add up to …” | `if target - x in seen` |
| Counting | “frequency”, “anagram”, “most common” | `Counter(items)` |
| Group by key | “group”, “categorise” | `groups[key(x)].append(x)` |
| Prefix sum + map | “subarray with sum k” (negatives allowed) | `count += seen[prefix - k]` |

## Challenge: Longest Consecutive Sequence: sort → set

Return the length of the longest run of consecutive integers in an unsorted list, e.g. `[100, 4, 200, 1, 3, 2]` → 4.

**Brute force** (O(n³)): For input like [1, 2, …, n]: each `in nums` scans O(n), the while loop runs up to n times per x, and there are n values of x. That is O(n³) in the worst case.

````python
def longest_consecutive(nums):
    best = 0
    for x in nums:
        length = 1
        while x + length in nums:        # list search
            length += 1
        best = max(best, length)
    return best
````

**Optimized** (O(n)): The inner while loop only runs from run starts, so across all x it advances at most n times in total.

A set gives O(1) membership, and a smart “only start here” condition stops you repeating work.

````python
def longest_consecutive(nums):
    s = set(nums)
    best = 0
    for x in s:
        if x - 1 not in s:
            length = 1
            while x + length in s:
                length += 1
            best = max(best, length)
    return best
````
