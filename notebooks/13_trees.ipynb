## 3. Depth-first traversals

Preorder, inorder and postorder all go deep first. The only difference is when the node itself is visited: before, between or after its subtrees.

| Traversal | Order | Typical use |
|---|---|---|
| Preorder | node, left, right | copy / serialize a tree |
| Inorder | left, node, right | sorted order of a BST |
| Postorder | left, right, node | delete a tree, compute sizes and heights bottom-up |

## 5. Thinking recursively about trees

Almost every tree problem has the same shape: handle None, ask both subtrees for their answer, combine with the current node.

**The template**

```python
def solve(node):
    if node is None:
        return BASE_ANSWER          # e.g. 0, True, []
    left = solve(node.left)         # trust the recursion
    right = solve(node.right)
    return combine(node.val, left, right)
```

## Challenge: Balanced Tree: height at every node → one bottom-up pass

A tree is balanced if, at every node, the left and right heights differ by at most 1.

**Brute force** (O(n²)): On a line-shaped tree, height() walks almost the whole tree for every node.

````python
def height(node):
    return 0 if node is None else 1 + max(height(node.left), height(node.right))

def is_balanced(node):
    if node is None:
        return True
    if abs(height(node.left) - height(node.right)) > 1:
        return False
    return is_balanced(node.left) and is_balanced(node.right)
````

**Optimized** (O(n)): Each node is visited once.

When a parent needs facts about its subtrees, return those facts from the recursion instead of recomputing them.

````python
def is_balanced(root):
    def check(node):              # height, or -1 if unbalanced
        if node is None:
            return 0
        l = check(node.left)
        r = check(node.right)
        if l == -1 or r == -1 or abs(l - r) > 1:
            return -1
        return 1 + max(l, r)
    return check(root) != -1
````
