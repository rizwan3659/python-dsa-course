## Challenge: Kth Smallest: sort everything → stop early in inorder

Return the k-th smallest value in a BST.

**Brute force** (O(n log n)): Visits every node, then sorts.

````python
def kth_smallest(root, k):
    vals = []
    def collect(n):
        if n:
            vals.append(n.val)
            collect(n.left)
            collect(n.right)
    collect(root)
    return sorted(vals)[k - 1]
````

**Optimized** (O(n)): Worst case still O(n), but typically O(h + k): it stops early and never sorts.

Structure is information: a BST is already sorted, so do not sort it again.

````python
def kth_smallest(root, k):
    stack, node = [], root
    while True:
        while node:
            stack.append(node)
            node = node.left
        node = stack.pop()
        k -= 1
        if k == 0:
            return node.val
        node = node.right
````
