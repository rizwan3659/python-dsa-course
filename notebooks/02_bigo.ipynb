## 1. Big-O in plain words

Big-O describes how the work grows as the input size n grows. We ignore constants and keep only the fastest-growing part.

- **Constant:** `def first(items):
    return items[0]`Index a list, look up a dict key, append to a list.
- **Logarithmic:** `while n > 1:
    n //= 2`Binary search: 1 million items in about 20 steps.
- **Linear:** `total = 0
for x in items:
    total += x`One loop over the input.
- **Linearithmic:** `sorted(items)`, merge sort. Usually the best you can do if you must sort.
- **Quadratic:** `for i in range(n):
    for j in range(n):
        ...`Compare every pair. 10,000 items → 100 million steps.
- **Exponential:** Naive recursive Fibonacci, all subsets. Unusable beyond n ≈ 30–40.

**Three rules**

- **Drop constants:** 3n + 5 → O(n). Two separate loops over n is still O(n).
- **Keep the biggest term:** n² + n → O(n²). For large n the smaller terms do not matter.
- **Nested multiplies, sequential adds:** A loop inside a loop: n × n. One loop after another: n + n.

## 5. What Python operations cost

Many one-liners hide a loop. Knowing these costs is how you spot an accidental O(n²).

| Operation | Average cost | Why |
|---|---|---|
| `lst[i]`, `lst[i] = x` | O(1) | Lists are arrays: jump straight to position i |
| `lst.append(x)`, `lst.pop()` | O(1) | Work at the end only (append is O(1) amortised: Python keeps spare room) |
| `lst.insert(0, x)`, `lst.pop(0)` | O(n) | Every other item shifts one place |
| `x in lst`, `lst.index(x)` | O(n) | Scans from the start |
| `lst[a:b]` | O(b − a) | Copies the slice |
| `sorted(lst)`, `lst.sort()` | O(n log n) | Timsort |
| `d[key]`, `key in d`, `d[key] = v` | O(1) | Hash table: jump straight to the slot for that key |
| `x in s`, `s.add(x)` | O(1) | Hash table |
| `deque.appendleft`, `popleft` | O(1) | Built for fast work at both ends |
| `s1 + s2` on strings | O(len) | Strings cannot change, so a new one is built |
| `"".join(parts)` | O(total length) | The right way to build a long string |
| `min(lst)`, `sum(lst)` | O(n) | Look at every item |

**See it: list vs set lookup**

```python
import time

n = 20000
items = list(range(n))
lookup = set(items)

start = time.perf_counter()
hits = sum(1 for x in range(n) if x in items)      # O(n) each → O(n²)
t_list = time.perf_counter() - start

start = time.perf_counter()
hits = sum(1 for x in range(n) if x in lookup)     # O(1) each → O(n)
t_set = time.perf_counter() - start

print(f"list: {t_list:.3f} s   set: {t_set:.4f} s   ~{t_list / max(t_set, 1e-9):.0f}x faster")
```

Timings vary by machine; the gap is what matters.

## Challenge: Contains Duplicate: all pairs → set

Return True if any value appears at least twice.

**Brute force** (O(n²)): Every pair is compared once.

```python
def has_duplicate(nums):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] == nums[j]:
                return True
    return False
```

**Optimized** (O(n)): One pass with O(1) work per element; O(n) extra space.

Same trick as Two Sum: an inner search becomes a set lookup.

```python
def has_duplicate(nums):
    seen = set()
    for x in nums:
        if x in seen:
            return True
        seen.add(x)
    return False
```
