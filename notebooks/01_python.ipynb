## 1. How to solve any coding problem

Most beginners start typing immediately. Strong problem solvers spend the first few minutes understanding and planning, and the code comes out right the first time.

**The UMPIRE method**

- **Understand:** Say the problem in your own words. What is the input? What is the output? Write 2 examples by hand, including an edge case (empty list, one element, negatives, duplicates).
- **Match:** Does it look like a known pattern? Counting → dictionary. Sorted input → two pointers or binary search. “All combinations” → recursion. You will collect these patterns throughout the course.
- **Plan:** Write the steps as comments before any code. If you cannot explain the plan, you are not ready to code it.
- **Implement:** Translate each comment into a few lines. Use clear names: `count`, not `c`.
- **Review:** Run your examples through the code in your head (or with `print`). Check the edge cases.
- **Evaluate:** State the time and space complexity (topic 2). Could it be faster?

## 3. Lists, tuples and slicing

A Python list is an ordered, changeable sequence. It is the array you will use most in this course. A tuple is the same idea, but cannot change.

**Lists: create, index, change**

```python
nums = [5, 3, 8, 1]
print(nums[0], nums[-1])     # first and last: 5 1
nums.append(9)               # add to the end
nums[1] = 30                 # change an item
print(nums, len(nums))       # [5, 30, 8, 1, 9] 5
print(sorted(nums))          # new sorted list; nums itself is unchanged
nums.sort(reverse=True)      # sorts nums in place
print(nums)
```

**Slicing: list[start:stop:step]**

```python
letters = ["a", "b", "c", "d", "e", "f"]
print(letters[1:4])    # ['b', 'c', 'd']  (stop is excluded)
print(letters[:3])     # first three
print(letters[-2:])    # last two
print(letters[::2])    # every second item
print(letters[::-1])   # reversed copy
point = (3, 4)         # a tuple: like a list, but cannot be changed
x, y = point           # unpacking
print(x + y)
```

## 4. Dicts and sets

A dictionary maps keys to values; a set holds unique items. Both look things up in about one step, which is why they appear in so many fast solutions.

**Dictionaries and sets**

```python
ages = {"asha": 21, "ravi": 19}
ages["meera"] = 22                 # add
print(ages.get("john", 0))         # default if missing: 0
for name, age in ages.items():
    print(name, age)

seen = set()
for x in [3, 1, 3, 2, 1]:
    seen.add(x)
print(seen, 2 in seen)             # {1, 2, 3} True
```

**Counting: the most common pattern in this course**

```python
from collections import Counter

text = "banana"
counts = {}
for ch in text:
    counts[ch] = counts.get(ch, 0) + 1
print(counts)                      # {'b': 1, 'a': 3, 'n': 2}
print(Counter(text).most_common(1))  # [('a', 3)]
```

## 6. Comprehensions, enumerate, zip and lambda

Short, readable tools you will use in almost every solution.

**Comprehensions, enumerate and zip**

```python
nums = [1, 2, 3, 4, 5, 6]
squares = [n * n for n in nums]
evens = [n for n in nums if n % 2 == 0]
print(squares, evens)

for i, value in enumerate(["x", "y", "z"]):
    print(i, value)

names, scores = ["a", "b"], [90, 75]
print(dict(zip(names, scores)))    # {'a': 90, 'b': 75}

pairs = [(2, "b"), (1, "z"), (2, "a")]
print(sorted(pairs, key=lambda p: (p[0], p[1])))   # sort by a key function
```

## 7. Classes: the building block for data structures

Every data structure you build from scratch (stack, linked list, tree…) is a class: data stored on self, plus methods that change it.

**A tiny class**

```python
class Node:
    def __init__(self, val, next=None):
        self.val = val          # data lives on the object
        self.next = next

    def __repr__(self):
        return f"Node({self.val})"

a = Node(1)
b = Node(2, a)                  # b.next points at a
print(b, b.next, b.next.next)   # Node(2) Node(1) None
```

## 8. Errors, copies and input

Real programs meet bad input, and real bugs come from sharing data you meant to copy. Handle the errors you expect, copy deliberately, and parse input in a few standard lines.

**try / except / else / finally**

```python
def to_int(text):
    try:
        value = int(text)
    except ValueError:            # only the error you expect
        print("not a number:", text)
        return None
    else:                         # runs when nothing was raised
        return value
    finally:                      # always runs (clean-up)
        print("checked", text)

print(to_int("42"), to_int("4x2"))
```

- **Deep copy:** `copy.deepcopy(a)` or `[row[:] for row in grid]` for a grid: everything new.

```python
import sys

def main(text):
    data = text.split()           # all tokens at once (fast)
    n = int(data[0])
    nums = list(map(int, data[1:1 + n]))
    print(sum(nums))

main("3\n4 5 6")                 # with real input: main(sys.stdin.read())
```
