## 5. Negative weights: Bellman-Ford

Dijkstra breaks when an edge can lower the cost of an already settled node. Bellman-Ford simply relaxes every edge, n − 1 times, and one extra round reveals negative cycles.

**Bellman-Ford**

```python
def bellman_ford(n, edges, source):
    INF = float("inf")
    dist = [INF] * n
    dist[source] = 0
    for _ in range(n - 1):                 # a shortest path has at most n - 1 edges
        for u, v, w in edges:
            if dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
    for u, v, w in edges:                  # still improving? negative cycle
        if dist[u] + w < dist[v]:
            raise ValueError("negative cycle")
    return dist

print(bellman_ford(4, [(0, 1, 4), (0, 2, 5), (1, 2, -3), (2, 3, 2)], 0))
```

## Challenge: Weighted shortest path: BFS → Dijkstra

Signals travel along weighted directed edges. How long until a signal from k reaches every node?

**Brute force** (O(n)): Fast (linear in V + E), but WRONG: with weights, the first path found is not necessarily the cheapest.

```python
from collections import deque

def network_delay(times, n, k):
    graph = {i: [] for i in range(1, n + 1)}
    for u, v, w in times:
        graph[u].append((v, w))
    dist, q = {k: 0}, deque([k])
    while q:
        u = q.popleft()
        for v, w in graph[u]:
            if v not in dist:
                dist[v] = dist[u] + w     # first arrival is final?!
                q.append(v)
    return max(dist.values()) if len(dist) == n else -1
```

**Optimized** (O(n log n)): O((V + E) log V) with a heap: a little slower than BFS, and correct.

“Optimal” first means correct. Pick the algorithm that matches the problem’s guarantees: BFS for unweighted, Dijkstra for non-negative weights, Bellman-Ford for negative weights.

```python
import heapq

def network_delay(times, n, k):
    graph = {i: [] for i in range(1, n + 1)}
    for u, v, w in times:
        graph[u].append((v, w))
    dist, pq = {k: 0}, [(0, k)]
    while pq:
        d, u = heapq.heappop(pq)
        if d > dist[u]:
            continue
        for v, w in graph[u]:
            if d + w < dist.get(v, float("inf")):
                dist[v] = d + w
                heapq.heappush(pq, (d + w, v))
    return max(dist.values()) if len(dist) == n else -1
```
