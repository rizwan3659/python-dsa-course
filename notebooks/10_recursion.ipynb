## 1. How recursion works

A recursive function solves a problem by solving a smaller copy of the same problem. Every recursive function needs a base case that stops it.

- **Base case:** The smallest input, answered directly. Without it the function calls itself forever (RecursionError).
- **Recursive case:** Call the function on a smaller input and use the result. Each call must move toward the base case.
- **Trust the call:** Assume `f(n - 1)` already works. Only think about how to get `f(n)` from it.
- **The call stack:** Each call waits on the stack until the smaller call returns. Python allows about 1,000 levels by default, so very deep recursion needs a loop instead.

## 2. The call stack and the return path

Calls go down until the base case. Then the answers travel back up, and each waiting call finishes its own work.

```text
 factorial(4)
 ↓
 4 × factorial(3)
       ↓
     3 × factorial(2)
           ↓
         2 × factorial(1)
               ↓
               1            ← base case
         2 × 1 = 2          ← return path
     3 × 2 = 6
 4 × 6 = 24
```

**Watch the stack grow and unwind**

````python
def factorial(n, depth=0):
    print("  " * depth + f"factorial({n}) called")
    if n <= 1:                         # base case
        print("  " * depth + "base case → 1")
        return 1
    result = n * factorial(n - 1, depth + 1)   # recursive case
    print("  " * depth + f"factorial({n}) returns {result}")
    return result

factorial(4)
````

## 3. Writing recursive functions

Three shapes cover most recursive functions: shrink by one, halve, and branch.

- **Process first, recurse on the rest:** `def total(nums):
    if not nums:
        return 0
    return nums[0] + total(nums[1:])`Clear, but slicing copies: O(n²) overall. Pass an index instead for O(n).
- **Divide the problem in two:** `def power(x, n):
    if n == 0:
        return 1
    half = power(x, n // 2)
    return half * half * (x if n % 2 else 1)`O(log n) multiplications.
- **Several recursive calls:** Fibonacci, tree traversal, generating subsets. Watch out for repeated work: memoize it.

## 4. Recursion trees and memoization

When a function calls itself more than once, the same inputs come back again and again. Cache them and exponential becomes linear.

**Visualizer:** Recursion tree and memoization: open https://rizwan3659.github.io/python-dsa-course/#v/fibtree

Set n = 6 and count the red boxes. Then turn memoization on.

**Memoization in one line**

````python
from functools import lru_cache

calls = 0

@lru_cache(maxsize=None)
def fib(n):
    global calls
    calls += 1
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(80), "calls:", calls)    # 81 calls instead of trillions
````

## 5. Backtracking

To generate every valid combination, make a choice, explore everything that follows from it, then undo the choice and try the next one.

**The template**

```python
def backtrack(state):
    if is_complete(state):
        record(copy_of(state))
        return
    for choice in choices(state):
        if is_valid(choice, state):
            make(choice, state)        # choose
            backtrack(state)           # explore
            undo(choice, state)        # un-choose
```

## 6. Pruning and constraint search

Backtracking explores an exponential tree, so the art is cutting branches early: stop when a partial answer already breaks a rule, sort so you can stop a loop, and skip choices that would repeat work.

- **Cut dead branches:** If the remaining sum is negative, a queen is attacked or a digit repeats, return immediately.
- **Sort to stop early:** In a sorted list, once a number is too big, every later number is too big as well: `break`.
- **Skip duplicates:** At one depth, try each distinct value once: `if i > start and nums[i] == nums[i - 1]: continue`.

## Challenge: Climbing Stairs: plain recursion → memoization

Taking 1 or 2 steps at a time, how many ways are there to climb n steps?

**Brute force** (O(2ⁿ)): Two calls per call: the same tree as naive Fibonacci.

````python
def climb(n):
    if n <= 2:
        return n
    return climb(n - 1) + climb(n - 2)
````

**Optimized** (O(n)): Each subproblem once; this bottom-up version even uses O(1) space.

Overlapping subproblems + caching = dynamic programming. You will meet it again in the Patterns topic.

````python
def climb(n):
    a, b = 1, 2                 # ways for 1 and 2 steps
    if n <= 2:
        return n
    for _ in range(3, n + 1):
        a, b = b, a + b
    return b
````
